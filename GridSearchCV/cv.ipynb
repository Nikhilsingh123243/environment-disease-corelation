```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv('loan_data.csv')
df.head()
```

```python
df.info()
```

```python
x = df.drop(columns=['loan_status'])
y = df['loan_status']
```

```python
num_cols = x.select_dtypes(include='number').columns
x = x[num_cols]
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,random_state=42,train_size=0.8)
```

```python
grid_search = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid={
           'criterion' : ['gini','entropy'],
           'splitter' : ['best','random'],
           'max_depth' : [None,5,10,20],
           'min_samples_split': [2,3,5,7],
           'min_samples_leaf' : [1,3,5,7],
           'max_features' : ['sqrt','log2']         
    },
    cv=3,
    verbose=1, # to get the information about fit we use verbose
    error_score='raise', #np.nan
    n_jobs=1 # it decides how many cpu cores to use. if -1 all cpu-cores and runs parallely.
)

grid_search.fit(xtrain,ytrain)
```

```python
grid_search.best_params_
```

```python
model = grid_search.best_estimator_
model.fit(xtrain,ytrain)
```

```python
results = pd.DataFrame(grid_search.cv_results_)
results.sort_values(by='rank_test_score')
```

```python
ytrain_pred = model.predict(xtrain) # For CV = 3

print(classification_report(ytrain,ytrain_pred))
```

```python
ytrain_pred = model.predict(xtrain) # For CV = 10

print(classification_report(ytrain,ytrain_pred))
```
