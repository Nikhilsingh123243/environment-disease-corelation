```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv('loan_data.csv')
df.head()
```

```python
df.info()
```

```python
x = df.drop(columns=['loan_status'])
y = df['loan_status']
```

```python
cat_cols = x.select_dtypes(include='number').columns
x = x[cat_cols]
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,random_state=42,train_size=0.8)
```

```python
preprocessing = ColumnTransformer(
    transformers = [
        ('encoder',OneHotEncoder(handle_unknown='ignore'),cat_cols)
    ],remainder='passthrough'
)
# preprocessing.fit_transformer(xtrain)
# preprocessing.get_feature_names_out()

main_pipeline = Pipeline(
    steps=[
        ('pre',preprocessing),
        ('model',DecisionTreeClassifier(random_state=42,criterion='entropy',max_depth=100))
    ]
)
grid_search = GridSearchCV(
    estimator=main_pipeline,
    param_grid={
        'model__criterion' : ['gini','entropy'],
        'model__splitter' : ['best','random'],
        'model__max_depth' : [None,5,10,20],
        'model__min_samples_split': [2,3,5,7],
        'model__min_samples_leaf' : [1,3,5,7],
        'model__max_features' : ['sqrt','log2']

    },
    cv=3
)

grid_search.fit(xtrain,ytrain)
```

```python
grid_search.best_params_
```

```python
model = grid_search.best_estimator_
model.fit(xtrain,ytrain)
```

```python
results = pd.DataFrame(grid_search.cv_results_)
results.sort_values(by='rank_test_score')
```

```python
ytrain_pred = model.predict(xtrain)

print(classification_report(ytrain,ytrain_pred))
```

```python
ytest_pred = model.predict(xtest)

print(classification_report(ytest,ytest_pred))
```
