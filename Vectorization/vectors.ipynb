```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('spam.csv',encoding='latin1')
df.head()
```

```python
df.info()
```

```python
x = df[['v2']]
y = df['v1']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,random_state=42,train_size=0.8)
```

```python
xtrain
```

```python
vectorizer = CountVectorizer(stop_words='english',binary=True,min_df=1,max_df=1.0,max_features=100,ngram_range=(2,2))

vectors = vectorizer.fit_transform(xtrain['v2']) #[['v2']] => 'v2' => 1
```
