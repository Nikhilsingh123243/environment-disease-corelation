```python
import pandas as pd
from sklearn.impute import SimpleImputer
import numpy as np

```

```python
df = pd.DataFrame({'DOM':['10','20','50',np.nan,'78'],
    'Product_type':['Electronics','Electronics',np.nan,'Acessories','Furniture'],
    'label':[1000,np.nan,300,400,650]
})
```

```python
df
```

```python
imputer = SimpleImputer(strategy="constant",fill_value=0)
```

```python
imputer.fit_transform(df[['DOM','Product_type']])
```

```python
imputer2 = SimpleImputer(strategy="constant",fill_value='unknown')
```

```python
imputer2.fit_transform(df[["DOM","label"]])
```

```python
imputer3 = SimpleImputer(strategy="mean")
```

```python
imputer3.fit_transform(df[["label"]])
```

```python
imputer4 = SimpleImputer(strategy="median")
imputer4.fit_transform(df[["DOM"]])
```

```python
imputer5 = SimpleImputer(strategy="most_frequent")
imputer5.fit_transform(df[["DOM","label"]])
```

```python
imputer6 = SimpleImputer(strategy="constant",fill_value=0)
imputer6.fit_transform(df[["DOM","label"]])
```

```python
imputer7 = SimpleImputer(strategy="most_frequent")
imputer7.fit_transform(df[["Product_type"]])
```
